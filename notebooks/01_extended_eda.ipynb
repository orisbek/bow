```python
# ============================================================
# WEEK 1 — SMS SPAM DETECTION
# EXTENDED EDA & DATA QUALITY
# ============================================================

# ============================================================
# 0. IMPORTS AND CONFIGURATION
# ============================================================

from pathlib import Path
from collections import Counter
import re
import unicodedata
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from IPython.display import display

warnings.filterwarnings("ignore")

pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", 200)

sns.set_theme(style="whitegrid")

# Change this path if necessary.
DATA_PATH = Path("../data/SMSSpamCollection")
# Alternative examples:
# DATA_PATH = Path("data/SMSSpamCollection")
# DATA_PATH = Path("/content/SMSSpamCollection")

RANDOM_STATE = 42

EXPECTED_LABELS = {"ham", "spam"}


# ============================================================
# 1.1. LOAD SMS SPAM COLLECTION
# ============================================================

if not DATA_PATH.exists():
    raise FileNotFoundError(
        f"Dataset was not found: {DATA_PATH.resolve()}\n"
        "Please update DATA_PATH with the correct file location."
    )

df = pd.read_csv(
    DATA_PATH,
    sep="\t",
    header=None,
    names=["label", "message"],
    encoding="utf-8",
    dtype={"label": "string", "message": "string"},
    keep_default_na=False
)

print("Dataset loaded successfully.")
print(f"Dataset path: {DATA_PATH.resolve()}")
print(f"Shape: {df.shape}")

display(df.head())


# ============================================================
# 1.2. STUDY DATASET STRUCTURE
# ============================================================

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
display(df.dtypes.to_frame("dtype"))

print("\nDataset info:")
df.info()


# ============================================================
# 1.3. CHECK DATASET FORMAT
# ============================================================

print("Expected columns:", ["label", "message"])
print("Actual columns:", df.columns.tolist())

assert list(df.columns) == ["label", "message"], (
    "Unexpected column names."
)

print("\nNumber of columns:", df.shape[1])
print("Number of rows:", df.shape[0])

print("\nMissing values:")
display(df.isna().sum().to_frame("missing_count"))

print("\nExample records:")
display(df.sample(min(5, len(df)), random_state=RANDOM_STATE))


# ============================================================
# 1.4. FEATURE DICTIONARY
# ============================================================

feature_dictionary = pd.DataFrame({
    "feature": ["label", "message"],
    "data_type": [
        str(df["label"].dtype),
        str(df["message"].dtype)
    ],
    "description": [
        "Target class: ham or spam",
        "Original SMS message text"
    ],
    "role": [
        "Target variable",
        "Input text"
    ]
})

display(feature_dictionary)


# ============================================================
# 2.1. CHECK MISSING VALUES
# ============================================================

missing_summary = pd.DataFrame({
    "missing_count": df.isna().sum(),
    "missing_percent": df.isna().mean() * 100
})

display(missing_summary)

print("Total missing values:", int(df.isna().sum().sum()))


# ============================================================
# 2.2. CHECK EMPTY AND WHITESPACE-ONLY MESSAGES
# ============================================================

message_as_string = df["message"].astype("string")

empty_message_mask = message_as_string.str.strip().eq("")

empty_message_count = int(empty_message_mask.sum())

print("Empty or whitespace-only messages:", empty_message_count)

if empty_message_count > 0:
    display(df.loc[empty_message_mask])
else:
    print("No empty or whitespace-only messages found.")


# ============================================================
# 2.3. CHECK TARGET LABELS
# ============================================================

unique_labels = set(df["label"].dropna().unique())

print("Unique labels:", unique_labels)

unexpected_labels = unique_labels - EXPECTED_LABELS

print("Unexpected labels:", unexpected_labels)

label_check = pd.DataFrame({
    "label": sorted(unique_labels),
    "is_expected": [
        label in EXPECTED_LABELS for label in sorted(unique_labels)
    ]
})

display(label_check)

if unexpected_labels:
    print("WARNING: Unexpected labels were found.")
else:
    print("All labels belong to the expected set.")


# ============================================================
# 2.4. CHECK DUPLICATES
# ============================================================

full_duplicate_count = int(df.duplicated().sum())
message_duplicate_count = int(df["message"].duplicated().sum())

print("Full duplicate rows:", full_duplicate_count)
print("Duplicate messages:", message_duplicate_count)

duplicate_messages = (
    df[df["message"].duplicated(keep=False)]
    .sort_values("message")
)

if not duplicate_messages.empty:
    print("\nExamples of duplicate messages:")
    display(duplicate_messages.head(20))
else:
    print("No duplicate messages found.")


# ============================================================
# DUPLICATES BEFORE AND AFTER HYPOTHETICAL REMOVAL
# ============================================================

df_without_full_duplicates = df.drop_duplicates().copy()
df_without_message_duplicates = df.drop_duplicates(
    subset=["message"],
    keep="first"
).copy()

duplicate_comparison = pd.DataFrame({
    "dataset_version": [
        "Original",
        "After full-row deduplication",
        "After message deduplication"
    ],
    "rows": [
        len(df),
        len(df_without_full_duplicates),
        len(df_without_message_duplicates)
    ],
    "removed_rows": [
        0,
        len(df) - len(df_without_full_duplicates),
        len(df) - len(df_without_message_duplicates)
    ]
})

display(duplicate_comparison)


# ============================================================
# 2.5. CHECK CONFLICTING LABELS
# ============================================================

label_counts_per_message = (
    df.groupby("message")["label"]
    .nunique()
)

conflicting_messages = label_counts_per_message[
    label_counts_per_message > 1
].index

conflicting_rows = df[
    df["message"].isin(conflicting_messages)
].sort_values("message")

print("Number of messages with conflicting labels:",
      len(conflicting_messages))

if len(conflicting_messages) > 0:
    display(conflicting_rows)
else:
    print("No conflicting labels found.")


# ============================================================
# 2.6. CHECK ANOMALOUS RECORDS
# ============================================================

df["message_length_chars"] = df["message"].str.len()
df["message_length_words"] = (
    df["message"]
    .str.split()
    .str.len()
)

print("Character length statistics:")
display(df["message_length_chars"].describe())

print("\nWord length statistics:")
display(df["message_length_words"].describe())

# Very long messages: inspect using the 99th percentile.
char_length_threshold = df["message_length_chars"].quantile(0.99)

very_long_messages = df[
    df["message_length_chars"] >= char_length_threshold
].sort_values("message_length_chars", ascending=False)

print(
    f"99th percentile character-length threshold: "
    f"{char_length_threshold:.2f}"
)

display(
    very_long_messages[
        ["label", "message", "message_length_chars"]
    ].head(20)
)

# Check unusual Unicode categories.
def get_unicode_category_counts(text):
    categories = Counter()

    for character in str(text):
        categories[unicodedata.category(character)] += 1

    return categories


def contains_control_character(text):
    return any(
        unicodedata.category(character).startswith("C")
        and character not in "\n\t\r"
        for character in str(text)
    )


df["contains_control_character"] = df["message"].apply(
    contains_control_character
)

control_character_count = int(
    df["contains_control_character"].sum()
)

print("Messages with possible control characters:",
      control_character_count)

if control_character_count > 0:
    display(
        df.loc[
            df["contains_control_character"],
            ["label", "message"]
        ].head(20)
    )


# ============================================================
# 3.1. COUNT HAM AND SPAM
# ============================================================

class_counts = df["label"].value_counts()

print("Absolute class counts:")
display(class_counts.to_frame("count"))

print("Total messages:", len(df))


# ============================================================
# 3.2. CALCULATE CLASS PERCENTAGES
# ============================================================

class_percentages = (
    df["label"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

class_distribution = pd.DataFrame({
    "count": class_counts,
    "percentage": class_percentages
})

display(class_distribution)


# ============================================================
# 3.3. EVALUATE CLASS IMBALANCE
# ============================================================

if set(class_counts.index) == EXPECTED_LABELS:
    majority_class = class_counts.idxmax()
    minority_class = class_counts.idxmin()

    majority_count = int(class_counts.max())
    minority_count = int(class_counts.min())

    imbalance_ratio = majority_count / minority_count

    print("Majority class:", majority_class)
    print("Minority class:", minority_class)
    print("Majority count:", majority_count)
    print("Minority count:", minority_count)
    print(f"Majority/minority ratio: {imbalance_ratio:.3f}")

    print(
        "\nInterpretation:"
        "\nAccuracy should not be used as the only metric."
        "\nPrecision, Recall, F1-score and the confusion matrix "
        "should also be evaluated."
    )
else:
    print(
        "Class imbalance analysis requires exactly ham and spam labels."
    )


# ============================================================
# 4.1. MESSAGE LENGTH IN CHARACTERS
# ============================================================

# Original text remains unchanged.
# This feature is used for EDA only.
df["message_length_chars"] = df["message"].str.len()

display(
    df[
        ["label", "message", "message_length_chars"]
    ].head()
)


# ============================================================
# 4.2. MESSAGE LENGTH IN WORDS
# ============================================================

# Word-count rule:
# split on whitespace after converting the message to string.
df["message_length_words"] = (
    df["message"]
    .astype("string")
    .str.split()
    .str.len()
)

display(
    df[
        ["label", "message", "message_length_words"]
    ].head()
)


# ============================================================
# 4.3. DESCRIPTIVE STATISTICS
# ============================================================

length_statistics = df[
    ["message_length_chars", "message_length_words"]
].describe(
    percentiles=[0.25, 0.50, 0.75, 0.95]
).T

display(length_statistics)


# ============================================================
# 4.4. COMPARE HAM AND SPAM MESSAGE LENGTH
# ============================================================

length_by_class = (
    df.groupby("label")[
        ["message_length_chars", "message_length_words"]
    ]
    .agg([
        "count",
        "min",
        "mean",
        "median",
        "std",
        "max"
    ])
    .round(2)
)

display(length_by_class)


# ============================================================
# 4.5. IDENTIFY EXTREMELY SHORT AND LONG MESSAGES
# ============================================================

shortest_messages = df.sort_values(
    "message_length_chars",
    ascending=True
).head(10)

longest_messages = df.sort_values(
    "message_length_chars",
    ascending=False
).head(10)

print("Shortest messages:")
display(
    shortest_messages[
        ["label", "message", "message_length_chars"]
    ]
)

print("Longest messages:")
display(
    longest_messages[
        ["label", "message", "message_length_chars"]
    ]
)


# ============================================================
# 5.1. COUNT UNIQUE MESSAGES
# ============================================================

total_messages = len(df)
unique_messages = df["message"].nunique()
duplicate_message_rows = total_messages - unique_messages

unique_message_summary = pd.DataFrame({
    "metric": [
        "Total rows",
        "Unique messages",
        "Rows beyond unique messages"
    ],
    "value": [
        total_messages,
        unique_messages,
        duplicate_message_rows
    ]
})

display(unique_message_summary)


# ============================================================
# 5.2. PRELIMINARY VOCABULARY ANALYSIS
# ============================================================

# This is a simple EDA tokenization rule:
# lowercase text and extract word-like sequences.
# It is not the final preprocessing pipeline.

def simple_tokenize(text):
    return re.findall(r"\b\w+\b", str(text).lower())


df["eda_tokens"] = df["message"].apply(simple_tokenize)
df["eda_token_count"] = df["eda_tokens"].str.len()

all_tokens = [
    token
    for tokens in df["eda_tokens"]
    for token in tokens
]

total_token_count = len(all_tokens)
unique_token_count = len(set(all_tokens))

vocabulary_summary = pd.DataFrame({
    "metric": [
        "Total tokens",
        "Unique tokens",
        "Vocabulary-to-token ratio"
    ],
    "value": [
        total_token_count,
        unique_token_count,
        unique_token_count / total_token_count
        if total_token_count > 0 else np.nan
    ]
})

display(vocabulary_summary)

vocabulary_by_class = {}

for label in sorted(df["label"].dropna().unique()):
    class_tokens = [
        token
        for tokens in df.loc[
            df["label"] == label, "eda_tokens"
        ]
        for token in tokens
    ]

    vocabulary_by_class[label] = {
        "messages": int((df["label"] == label).sum()),
        "total_tokens": len(class_tokens),
        "unique_tokens": len(set(class_tokens))
    }

vocabulary_by_class_df = pd.DataFrame(vocabulary_by_class).T

display(vocabulary_by_class_df)


# ============================================================
# 5.3. NUMERIC AND SYMBOLIC PATTERNS
# ============================================================

def count_pattern(series, pattern):
    return series.astype("string").str.contains(
        pattern,
        regex=True,
        case=False,
        na=False
    )


# URL-like patterns
df["contains_url"] = count_pattern(
    df["message"],
    r"(https?://|www\.|bit\.ly/|t\.co/)"
)

# Any digit
df["contains_digit"] = count_pattern(
    df["message"],
    r"\d"
)

# Currency symbols and common currency abbreviations
df["contains_currency"] = count_pattern(
    df["message"],
    r"[$€£¥₹]|\b(?:usd|eur|gbp)\b"
)

# Exclamation mark
df["contains_exclamation"] = df["message"].str.contains(
    "!",
    regex=False,
    na=False
)

# Uppercase Latin letters
df["contains_uppercase"] = df["message"].str.contains(
    r"[A-Z]",
    regex=True,
    na=False
)

# Repeated punctuation, e.g. !!! or ???
df["contains_repeated_punctuation"] = count_pattern(
    df["message"],
    r"([!?.,])\1{1,}"
)

# Non-ASCII characters
df["contains_non_ascii"] = df["message"].apply(
    lambda text: any(ord(character) > 127 for character in str(text))
)

pattern_columns = [
    "contains_url",
    "contains_digit",
    "contains_currency",
    "contains_exclamation",
    "contains_uppercase",
    "contains_repeated_punctuation",
    "contains_non_ascii"
]

pattern_summary = pd.DataFrame({
    "count": [int(df[column].sum()) for column in pattern_columns],
    "percentage": [
        round(float(df[column].mean() * 100), 2)
        for column in pattern_columns
    ]
}, index=pattern_columns)

display(pattern_summary)


# ============================================================
# 5.4. COMPARE SURFACE FEATURES BY CLASS
# ============================================================

surface_features_by_class = (
    df.groupby("label")[pattern_columns]
    .agg(["sum", "mean"])
    .round(4)
)

display(surface_features_by_class)

# More readable percentage table.
surface_percentage_by_class = (
    df.groupby("label")[pattern_columns]
    .mean()
    .mul(100)
    .round(2)
)

print("Percentage of messages containing each pattern:")
display(surface_percentage_by_class)


# ============================================================
# ADDITIONAL EDA: TOP TOKENS BY CLASS
# ============================================================

def top_tokens_for_class(dataframe, label, top_n=20):
    tokens = [
        token
        for tokens in dataframe.loc[
            dataframe["label"] == label, "eda_tokens"
        ]
        for token in tokens
    ]

    return pd.DataFrame(
        Counter(tokens).most_common(top_n),
        columns=["token", "frequency"]
    )


for label in sorted(df["label"].dropna().unique()):
    print(f"\nTop tokens for class: {label.upper()}")
    display(top_tokens_for_class(df, label, top_n=20))


# ============================================================
# ADDITIONAL EDA: COMMON BIGRAMS
# ============================================================

def make_ngrams(tokens, n=2):
    return [
        tuple(tokens[index:index + n])
        for index in range(len(tokens) - n + 1)
    ]


def top_ngrams_for_class(dataframe, label, n=2, top_n=20):
    ngrams = []

    class_token_lists = dataframe.loc[
        dataframe["label"] == label, "eda_tokens"
    ]

    for tokens in class_token_lists:
        ngrams.extend(make_ngrams(tokens, n=n))

    frequencies = Counter(ngrams).most_common(top_n)

    return pd.DataFrame({
        "ngram": [" ".join(item[0]) for item in frequencies],
        "frequency": [item[1] for item in frequencies]
    })


for label in sorted(df["label"].dropna().unique()):
    print(f"\nTop bigrams for class: {label.upper()}")
    display(top_ngrams_for_class(df, label, n=2, top_n=20))


# ============================================================
# 6.1. VISUALIZATIONS
# ============================================================

# ------------------------------------------------------------
# Plot 1: Class distribution
# ------------------------------------------------------------

plt.figure(figsize=(7, 5))

ax = sns.countplot(
    data=df,
    x="label",
    order=sorted(df["label"].dropna().unique())
)

plt.title("Class Distribution: HAM vs SPAM")
plt.xlabel("Class")
plt.ylabel("Number of messages")

for container in ax.containers:
    ax.bar_label(container)

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# Plot 2: Character-length distribution
# ------------------------------------------------------------

plt.figure(figsize=(10, 5))

sns.histplot(
    data=df,
    x="message_length_chars",
    hue="label",
    bins=50,
    kde=True,
    element="step",
    stat="density",
    common_norm=False
)

plt.title("Message Length Distribution by Class")
plt.xlabel("Message length in characters")
plt.ylabel("Density")

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# Plot 3: Word-count distribution
# ------------------------------------------------------------

plt.figure(figsize=(10, 5))

sns.histplot(
    data=df,
    x="message_length_words",
    hue="label",
    bins=40,
    kde=True,
    element="step",
    stat="density",
    common_norm=False
)

plt.title("Word Count Distribution by Class")
plt.xlabel("Number of words")
plt.ylabel("Density")

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# Plot 4: Boxplot of character length
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="label",
    y="message_length_chars",
    order=sorted(df["label"].dropna().unique())
)

plt.title("Character Length by Class")
plt.xlabel("Class")
plt.ylabel("Message length in characters")

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# Plot 5: Boxplot of word count
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="label",
    y="message_length_words",
    order=sorted(df["label"].dropna().unique())
)

plt.title("Word Count by Class")
plt.xlabel("Class")
plt.ylabel("Number of words")

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# Plot 6: Surface-feature percentages
# ------------------------------------------------------------

surface_plot_data = (
    df.groupby("label")[pattern_columns]
    .mean()
    .mul(100)
    .T
)

plt.figure(figsize=(12, 6))

surface_plot_data.plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title("Surface Features by Class")
plt.xlabel("Feature")
plt.ylabel("Percentage of messages")
plt.xticks(rotation=45, ha="right")
plt.legend(title="Class")

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# Plot 7: Duplicate summary
# ------------------------------------------------------------

duplicate_plot_data = pd.DataFrame({
    "category": [
        "Total rows",
        "Unique messages",
        "Duplicate message rows"
    ],
    "count": [
        total_messages,
        unique_messages,
        duplicate_message_rows
    ]
})

plt.figure(figsize=(8, 5))

sns.barplot(
    data=duplicate_plot_data,
    x="category",
    y="count"
)

plt.title("Message Uniqueness Summary")
plt.xlabel("")
plt.ylabel("Count")
plt.xticks(rotation=20)

plt.tight_layout()
plt.show()


# ============================================================
# 6.2. FINAL EDA RESULTS TABLE
# ============================================================

quality_results = pd.DataFrame({
    "check": [
        "Total rows",
        "Total columns",
        "Missing values",
        "Empty messages",
        "Unexpected labels",
        "Full duplicate rows",
        "Duplicate messages",
        "Conflicting labels",
        "Messages with control characters"
    ],
    "value": [
        len(df),
        df.shape[1],
        int(df.isna().sum().sum()),
        empty_message_count,
        len(unexpected_labels),
        full_duplicate_count,
        message_duplicate_count,
        len(conflicting_messages),
        control_character_count
    ]
})

print("Data quality summary:")
display(quality_results)

eda_summary = pd.DataFrame({
    "metric": [
        "Total messages",
        "Unique messages",
        "HAM messages",
        "SPAM messages",
        "HAM percentage",
        "SPAM percentage",
        "Mean character length",
        "Median character length",
        "Mean word count",
        "Median word count",
        "Unique EDA tokens",
        "Total EDA tokens"
    ],
    "value": [
        len(df),
        unique_messages,
        int((df["label"] == "ham").sum()),
        int((df["label"] == "spam").sum()),
        round(float((df["label"] == "ham").mean() * 100), 2),
        round(float((df["label"] == "spam").mean() * 100), 2),
        round(float(df["message_length_chars"].mean()), 2),
        round(float(df["message_length_chars"].median()), 2),
        round(float(df["message_length_words"].mean()), 2),
        round(float(df["message_length_words"].median()), 2),
        unique_token_count,
        total_token_count
    ]
})

print("EDA summary:")
display(eda_summary)


# ============================================================
# 6.3. AUTOMATICALLY GENERATE EDA CONCLUSIONS
# ============================================================

print("=" * 70)
print("EDA CONCLUSIONS")
print("=" * 70)

print(f"1. The dataset contains {len(df):,} rows and {df.shape[1]} columns.")

print(
    f"2. The dataset contains "
    f"{int((df['label'] == 'ham').sum()):,} HAM messages and "
    f"{int((df['label'] == 'spam').sum()):,} SPAM messages."
)

print(
    f"3. There are {unique_messages:,} unique messages and "
    f"{duplicate_message_rows:,} rows beyond unique messages."
)

print(
    f"4. The average message length is "
    f"{df['message_length_chars'].mean():.2f} characters."
)

print(
    f"5. The median message length is "
    f"{df['message_length_chars'].median():.2f} characters."
)

print(
    f"6. The average message word count is "
    f"{df['message_length_words'].mean():.2f}."
)

print(
    f"7. Messages containing URLs: "
    f"{int(df['contains_url'].sum()):,} "
    f"({df['contains_url'].mean() * 100:.2f}%)."
)

print(
    f"8. Messages containing digits: "
    f"{int(df['contains_digit'].sum()):,} "
    f"({df['contains_digit'].mean() * 100:.2f}%)."
)

print(
    f"9. Messages containing currency symbols or abbreviations: "
    f"{int(df['contains_currency'].sum()):,} "
    f"({df['contains_currency'].mean() * 100:.2f}%)."
)

if len(conflicting_messages) > 0:
    print(
        f"10. Conflicting labels were found in "
        f"{len(conflicting_messages):,} unique messages. "
        f"These records require manual inspection."
    )
else:
    print(
        "10. No conflicting labels were found."
    )

print(
    "\nPreprocessing decisions to investigate in Week 2:"
)

print(
    "- Compare minimal cleaning with more aggressive cleaning."
)
print(
    "- Test whether numbers and currency symbols should be retained."
)
print(
    "- Test whether URLs should be replaced with a special token "
    "instead of simply removed."
)
print(
    "- Compare tokenization and normalization strategies."
)
print(
    "- Avoid fitting vocabulary-based transformations on the full "
    "dataset before train/test splitting."
)


# ============================================================
# OPTIONAL: SAVE EDA OUTPUTS
# ============================================================

OUTPUT_DIR = Path("../reports/eda")

# Set SAVE_OUTPUTS = True if you want to export tables.
SAVE_OUTPUTS = False

if SAVE_OUTPUTS:
    OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

    quality_results.to_csv(
        OUTPUT_DIR / "data_quality_summary.csv",
        index=False
    )

    eda_summary.to_csv(
        OUTPUT_DIR / "eda_summary.csv",
        index=False
    )

    class_distribution.to_csv(
        OUTPUT_DIR / "class_distribution.csv"
    )

    length_statistics.to_csv(
        OUTPUT_DIR / "length_statistics.csv"
    )

    surface_percentage_by_class.to_csv(
        OUTPUT_DIR / "surface_features_by_class.csv"
    )

    print(f"EDA tables saved to: {OUTPUT_DIR.resolve()}")


# ============================================================
# END OF WEEK 1
# ============================================================
```
